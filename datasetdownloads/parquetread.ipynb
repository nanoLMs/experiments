This notebook demonstrates how to read and work with Parquet datasets using pandas and pyarrow.

```python
# Import required libraries
import pandas as pd
import pyarrow.parquet as pq
```

```python
# Function to read a parquet file
def read_parquet_file(file_path):
    try:
        # Read the parquet file into a pandas DataFrame
        df = pd.read_parquet(file_path)

        # Print basic information about the dataset
        print("Dataset Info:")
        print("-" * 50)
        print("\nDataset Shape:", df.shape)
        print("\nColumns:", df.columns.tolist())
        print("\nData Types:")
        print(df.dtypes)
        print("\nFirst few rows:")
        print(df.head())

        return df

    except Exception as e:
        print(f"Error reading parquet file: {str(e)}")
        return None

# Example usage:
# Replace 'path_to_your_parquet_file.parquet' with your actual parquet file path
# df = read_parquet_file('path_to_your_parquet_file.parquet')
```
